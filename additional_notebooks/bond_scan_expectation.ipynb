```python
import json
import matplotlib.pyplot as plt
import numpy as np

with open('../data/h2_bond_scan_expectation.json', 'r') as f:
    data_bond_scan = json.load(f)

bond_scan_distances = data_bond_scan['distances']
bond_scan_energies = data_bond_scan['energies']

plt.figure(figsize=(12, 7))
plt.plot(bond_scan_distances, bond_scan_energies, 'b-o', linewidth=5, markersize=10, label='Bond Scan', alpha = 1, color='royalblue', zorder=1)

# plt.ylim(-1.2, -0.1)
# for close up image
# plt.ylim(-1.15, -1.1)
# plt.xlim(0.64, 0.84)

plt.xlabel('Distance / Å', fontsize=20, labelpad=15)
plt.ylabel('Energy / Hartree', fontsize=20, labelpad=15)
plt.title('H$_2$ Bond Scan Estimation', fontsize=24, fontweight='bold', pad=20)
# plt.legend(fontsize=16, frameon=False)
plt.grid(True, alpha=0.7, linewidth=1.5)
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(2)
ax.tick_params(axis='both', which='major', labelsize=16, length=8, width=2)
ax.set_axisbelow(True)  # Put grid behind plot elements

plt.tight_layout()

# plt.savefig('../images/presentation/h2_bond_scan_expectation.png', dpi=300, bbox_inches='tight')
plt.show()
```
