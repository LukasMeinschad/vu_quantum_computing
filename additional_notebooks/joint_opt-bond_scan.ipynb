```python
import json
import matplotlib.pyplot as plt
import numpy as np

with open('../data/h2_bond_scan_exact.json', 'r') as f:
    data_bond_scan = json.load(f)

with open('../data/h2_joint_optimization_exact.json', 'r') as f:
    data_joint_opt = json.load(f)

bond_scan_distances = data_bond_scan['distances']
bond_scan_energies = data_bond_scan['energies']

joint_opt_distances = data_joint_opt['history_distances']
joint_opt_energies = data_joint_opt['history_raw_energies']
optimal_distance = data_joint_opt['optimal_distance']
optimal_energy = data_joint_opt['optimal_energy']

plt.figure(figsize=(12, 7))
plt.plot(bond_scan_distances, bond_scan_energies, 'b-o', linewidth=3, markersize=6, label='Bond Scan', alpha = 0.8, color='navy')
plt.plot(joint_opt_distances, joint_opt_energies, 'g-s', linewidth=3, markersize=6, label='Joint Optimization', alpha = 0.8, color='darkorange')
plt.scatter(optimal_distance, optimal_energy, color='red', s=200, marker='*', 
            label=f'Optimal Distance', 
            edgecolors='black', linewidth=1.5, zorder=10)

plt.ylim(-1.2, -0.1)
# for close up image
# plt.ylim(-1.15, -1.1)
# plt.xlim(0.64, 0.84)

plt.xlabel('Distance / Å', fontsize=20, labelpad=15)
plt.ylabel('Energy / Hartree', fontsize=20, labelpad=15)
plt.title('Bond Scan vs Joint Optimization', fontsize=24, fontweight='bold', pad=20)
plt.legend(fontsize=16, frameon=False)
plt.grid(True, alpha=0.7, linewidth=1.5)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(2)
ax.tick_params(axis='both', which='major', labelsize=16, length=8, width=2)

plt.tight_layout()

# plt.savefig('../images/presentation/h2_bond_scan_joint_opt_close_up.png', dpi=300, bbox_inches='tight')
plt.show()
```
