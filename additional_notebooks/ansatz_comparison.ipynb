```python
import json
import matplotlib.pyplot as plt
import numpy as np

# Load the two convergence data files
with open('../data/vqe_conv_efficientSU2_jw_2.json', 'r') as f:
    data_effsu2 = json.load(f)

with open('../data/vqe_conv_uccsd_jw_2.json', 'r') as f:
    data_uccsd = json.load(f)

# Extract convergence data
conv_effsu2 = data_effsu2['convergence'][0]
conv_uccsd = data_uccsd['convergence'][0]

# Print convergence statistics
print("EfficientSU2 Ansatz:")
print(f"  Final Energy: {conv_effsu2[-1]:.8f} Hartree")
print(f"  Total Iterations: {len(conv_effsu2)}")
print()
print("UCCSD Ansatz:")
print(f"  Final Energy: {conv_uccsd[-1]:.8f} Hartree")
print(f"  Total Iterations: {len(conv_uccsd)}")
print()

# Create the comparison plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(conv_effsu2, label='EfficientSU2', linewidth=2.5, marker='o', markersize=0, markevery=5, color='royalblue')
ax.plot(conv_uccsd, label='UCCSD', linewidth=2.5, marker='s', markersize=0, markevery=5, color='darkorange')

ax.set_xlabel('Cost Function Evaluations', fontsize=20, labelpad=15)
ax.set_ylabel('Energy / Hartree', fontsize=20 , labelpad=15)
ax.set_title('VQE Convergence Comparison: EfficientSU2 vs UCCSD', fontsize=22, fontweight='bold', pad=20)

ax.set_xlim(0, 200)

# Increase tick label size and tick length/thickness
ax.tick_params(axis='both', which='major', labelsize=16, length=8, width=2)

# Increase frame thickness
for spine in ax.spines.values():
    spine.set_linewidth(2)

ax.grid(True, alpha=0.7, linewidth=1.5)
ax.legend(fontsize=18, loc='best')
plt.tight_layout()

plt.savefig('../images/presentation/ansatz_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
