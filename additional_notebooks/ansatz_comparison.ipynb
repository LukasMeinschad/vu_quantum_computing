```python
import json
import matplotlib.pyplot as plt
import numpy as np

# Load the two convergence data files
with open('../data/vqe_conv_efficientSU2_jw_2_lih.json', 'r') as f:
    data_effsu2 = json.load(f)

with open('../data/vqe_conv_uccsd_jw_2_lih.json', 'r') as f:
    data_uccsd = json.load(f)

# Extract convergence data
conv_effsu2 = data_effsu2['convergence'][0]
conv_uccsd = data_uccsd['convergence'][0]

# Create the comparison plot
plt.figure(figsize=(10, 6))
plt.plot(conv_effsu2, label='EfficientSU2', linewidth=2.0, marker='o', markersize=4, markevery=5)
plt.plot(conv_uccsd, label='UCCSD', linewidth=2.0, marker='s', markersize=4, markevery=5)

plt.xlabel('Cost Function Evaluations', fontsize=12)
plt.ylabel('Energy (Ha)', fontsize=12)
plt.title('VQE Convergence Comparison: EfficientSU2 vs UCCSD', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11, loc='best')
plt.tight_layout()

# Print some statistics
print(f"EfficientSU2 - Final Energy: {conv_effsu2[-1]:.10f} Ha")
print(f"UCCSD - Final Energy: {conv_uccsd[-1]:.10f} Ha")
print(f"Energy Difference: {abs(conv_effsu2[-1] - conv_uccsd[-1]):.6e} Ha")
print(f"\nEfficientSU2 - Evaluations: {len(conv_effsu2)}")
print(f"UCCSD - Evaluations: {len(conv_uccsd)}")

plt.show()
```
