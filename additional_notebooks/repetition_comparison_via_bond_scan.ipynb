```python
import json
import matplotlib.pyplot as plt
import numpy as np

# Load the bond scan data files
with open('../data/h2_bond_scan_rep1.json', 'r') as f:
    data_1rep = json.load(f)
    
with open('../data/h2_bond_scan_rep2.json', 'r') as f:
    data_2rep = json.load(f)

with open('../data/h2_bond_scan_rep3.json', 'r') as f:
    data_3rep = json.load(f)
    
with open('../data/h2_bond_scan_rep4.json', 'r') as f:
    data_4rep = json.load(f)
    
with open('../data/h2_bond_scan_rep5.json', 'r') as f:
    data_5rep = json.load(f)

# Extract distances and energies
distances_1 = data_1rep['distances']
energies_1 = data_1rep['energies']

distances_2 = data_2rep['distances']
energies_2 = data_2rep['energies']

distances_3 = data_3rep['distances']
energies_3 = data_3rep['energies']

distances_4 = data_4rep['distances']
energies_4 = data_4rep['energies']

distances_5 = data_5rep['distances']
energies_5 = data_5rep['energies']

# Print minimum energies and corresponding bond lengths
print("Minimum Energies:")
print(f"  1 rep: {min(energies_1):.8f} Ha at {distances_1[energies_1.index(min(energies_1))]:.4f} Å")
print(f"  2 rep: {min(energies_2):.8f} Ha at {distances_2[energies_2.index(min(energies_2))]:.4f} Å")
print(f"  3 rep: {min(energies_3):.8f} Ha at {distances_3[energies_3.index(min(energies_3))]:.4f} Å")
print(f"  4 rep: {min(energies_4):.8f} Ha at {distances_4[energies_4.index(min(energies_4))]:.4f} Å")
print(f"  5 rep: {min(energies_5):.8f} Ha at {distances_5[energies_5.index(min(energies_5))]:.4f} Å")
print()

# Create the comparison plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(distances_1, energies_1, label='1 rep', linewidth=2.5, marker='o', markersize=6, color='royalblue')
ax.plot(distances_2, energies_2, label='2 rep', linewidth=2.5, marker='s', markersize=6, color='darkorange')
ax.plot(distances_3, energies_3, label='3 rep', linewidth=2.5, marker='^', markersize=6, color='green')
ax.plot(distances_4, energies_4, label='4 rep', linewidth=2.5, marker='D', markersize=6, color='red')
ax.plot(distances_5, energies_5, label='5 rep', linewidth=2.5, marker='v', markersize=6, color='purple')

ax.set_xlabel('Bond Distance / Å', fontsize=20, labelpad=15)
ax.set_ylabel('Energy / Hartree', fontsize=20, labelpad=15)
ax.set_title('H₂ Bond Scan: Number of Ansatz Repetitions', fontsize=22, fontweight='bold', pad=20)

# Increase tick label size and tick length/thickness
ax.tick_params(axis='both', which='major', labelsize=16, length=8, width=2)

# Increase frame thickness
for spine in ax.spines.values():
    spine.set_linewidth(2)

ax.grid(True, alpha=0.7, linewidth=1.5)
ax.legend(fontsize=18, loc='best')
plt.tight_layout()

# plt.savefig('../images/presentation/h2_bond_scan_repetition_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
