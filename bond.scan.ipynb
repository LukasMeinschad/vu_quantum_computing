## Configuration 

The following cell is the central configuration to set the number of shots and configure the Backend to be used. For testing reasons we used the AerSimulator.

**Calculation Setup**

We want to run a bond-scan for the hydrogen molecule in the range from 0.54Å to 0.94Å with 5 points. At each point we build a UCCSD ansatz with 5 reps, this adds up to 361 gates in total (see cell below for details).

+ Each VQE evaluation will be run with a total number of 20 optimization steps using COBYLA as a classical optimizer
+ The number of shots per expectation value is set to 200 
+ As seen in the last step of this notebook one VQE minimization for a specific bond length will use 56000 shots in total
+ This multiplies to a total shot count of 280000 for the whole bond scan with 5 points 


```python
BACKEND = AerSimulator() # Set the AQT backend here
SHOTS = 200 # Set the Number of Shots here 
OPT_LEVEL = 1 # Set the Optimization Level of the Circuit Compilation
sampler = SamplerV2()
```

The following cell makes the measurement circuit and estimates our operator based on shots and backend defined in the configuration cell above.

```python
def _build_measurement_circuit_for_pauli(base_circuit, pauli_label: str):
    """
    Measurment circuit for a given Pauli string.
    Basis change for X/Y terms and then measure_all().
    """
    qc = base_circuit.copy()
    n = len(pauli_label)
    if qc.num_qubits != n:
        raise ValueError(f"Pauli label length {n} != circuit qubits {qc.num_qubits}")

    for qubit in range(n):
        op = pauli_label[n - 1 - qubit]  # qubit-0 ist rechts
        if op == "I":
            continue
        if op == "X":
            qc.h(qubit)
        elif op == "Y":
            qc.sdg(qubit)
            qc.h(qubit)
        elif op == "Z":
            pass
        else:
            raise ValueError(f"Unsupported Pauli op: {op}")

    qc.measure_all()
    return qc


def _counts_to_pauli_expectation(counts: dict[str, int], pauli_label: str) -> float:
    """
    Converts Z-basis counts to <P>; assumes basis change has already been applied.
    """
    n = len(pauli_label)
    active_qubits = [q for q in range(n) if pauli_label[n - 1 - q] != "I"]
    if not active_qubits:
        return 1.0

    shots = int(sum(counts.values()))
    if shots <= 0:
        raise ValueError("No shots in counts")

    exp = 0.0
    for bitstring, c in counts.items():
        parity = 0
        for q in active_qubits:
            parity ^= (bitstring[n - 1 - q] == "1")
        exp += (-1.0 if parity else 1.0) * float(c)

    return exp / float(shots)


def sampler_expectation_sparsepauliop(
    *,
    sampler,
    base_circuit,
    params: np.ndarray,
    operator,          # SparsePauliOp
    shots: int,
):
    """
    Determines the expectation value of `operator` with respect to the state
    prepared by `base_circuit` with `params`, using the provided `sampler`
    """
    labels = list(operator.paulis.to_labels())
    coeffs = np.asarray(operator.coeffs)

    meas_circuits = []
    pub_params = []
    keep_idx = []

    energy = 0.0
    for i, (lab, coeff) in enumerate(zip(labels, coeffs)):
        coeff_r = float(np.real(coeff))
        if np.isclose(coeff_r, 0.0):
            continue
        if set(lab) == {"I"}:
            energy += coeff_r
            continue
        meas_circuits.append(_build_measurement_circuit_for_pauli(base_circuit, lab))
        pub_params.append(params)
        keep_idx.append(i)

    if meas_circuits:
        # SamplerV2: pubs = [(circuit, params), ...]
        pubs = list(zip(meas_circuits, pub_params))
        job = sampler.run(pubs, shots=int(shots))
        res = job.result()

        for j, i in enumerate(keep_idx):
            counts = res[j].data.meas.get_counts()
            term_exp = _counts_to_pauli_expectation(counts, labels[i])
            energy += float(np.real(coeffs[i])) * float(term_exp)

    return float(energy)
```

## Step 1: Build the Fermionic Hamiltonian parametrized by the bond length



```python
def build_h2_fermionic_hamiltonian(
    distance: float,
    *,
    basis: str = "sto3g",
    charge: int = 0,
    spin: int = 0,
    freeze_core: bool = False,
    active_space: tuple[int, int] | None = None,  # (num_electrons, num_spatial_orbitals)
):
    """
    Returns the fermionic Hamiltonian which is parametrized on the bond distance x
    """
    driver = PySCFDriver(
        atom=f"H 0 0 0; H 0 0 {float(distance)}",
        unit=DistanceUnit.ANGSTROM,
        basis=basis,
        charge=charge,
        spin=spin,
    )
    problem = driver.run()

    if freeze_core:
        problem = FreezeCoreTransformer().transform(problem)

    if active_space is not None:
        nelec, norb = map(int, active_space)
        problem = ActiveSpaceTransformer(
            num_electrons=nelec, num_spatial_orbitals=norb
        ).transform(problem)

    fermionic_ham = problem.second_q_ops()[0]
    return fermionic_ham, problem
```

## Step 2: Map the Fermionic Hamiltonian to the Qubit Hamiltonian

In this step, we use the Jordan-Wigner mapping to convert the Fermionic Hamiltonian into a Qubit Hamiltonian. Further transformations are tested in the original codebase on GitHub.

```python
def build_h2_qubit_hamiltonian_jw(
    distance: float,
    *,
    basis: str = "sto3g",
    charge: int = 0,
    spin: int = 0,
    freeze_core: bool = False,
    active_space: tuple[int, int] | None = None,
):
    """
    Builds a qubit Hamiltonian for H2 for a given distance using the jordan Wigner mapping 
    """
    fermionic_ham, problem = build_h2_fermionic_hamiltonian(
        distance,
        basis=basis,
        charge=charge,
        spin=spin,
        freeze_core=freeze_core,
        active_space=active_space,
    )
    qubit_ham = JordanWignerMapper().map(fermionic_ham)
    return qubit_ham, problem
```

# Step 3: Prepare the Ansatz Circuit

In the next step, we prepare the ansatz circuit using the UCCSD method with 5 reps.

```python
def build_h2_uccsd_ansatz(
    problem,
    *,
    reps: int = 5,
):
    """
    Builds a UCCSD ansatz for H2 using Jordan-Wigner + Hartree-Fock initial state.
    """
    mapper = JordanWignerMapper()

    hf = HartreeFock(
        num_spatial_orbitals=problem.num_spatial_orbitals,
        num_particles=problem.num_particles,
        qubit_mapper=mapper,
    )

    return UCCSD(
        num_spatial_orbitals=problem.num_spatial_orbitals,
        num_particles=problem.num_particles,
        qubit_mapper=mapper,
        initial_state=hf,
        reps=reps,
    )
Hq, problem = build_h2_qubit_hamiltonian_jw(0.74)
ansatz = build_h2_uccsd_ansatz(problem, reps=5)


# The following provides the gate count after the Transpile 
pm = generate_preset_pass_manager(backend=BACKEND, optimization_level=int(OPT_LEVEL))
isa_ansatz = pm.run(ansatz)

print("\nOps (nach Transpile):")
print(isa_ansatz.count_ops())
print("Depth (nach Transpile):", isa_ansatz.depth())
```

# Step 4: Set up the VQE algorithm with the COBYLA optimizer

```python
def interpret_total_energy(exp_val, problem) -> float:
    """
    Convert Estimator expectation value -> total energy (Ha) via problem.interpret.
    """
    ev = np.real(exp_val)
    if isinstance(ev, np.ndarray):
        ev = ev.item() if ev.size == 1 else float(ev.ravel()[0])

    sol = MinimumEigensolverResult()
    sol.eigenvalue = float(ev)
    return float(problem.interpret(sol).total_energies[0])

def run_vqe_single_point_sampler(
    *,
    problem,
    qubit_hamiltonian,
    ansatz,
    sampler,
    shots: int = 4096,
    backend=None,              
    optimization_level: int = 1,
    x0: np.ndarray | None = None,
    maxiter: int = 120,
    seed: int = 42,
    verbose: bool = False,
):
    """
    VQE Solver with shot based Sampling
    """

    if backend is None:
        backend = AerSimulator()

    # Transpile -> entfernt High-Level Ops wie EvolvedOps für Aer/Hardware
    pm = generate_preset_pass_manager(backend=backend, optimization_level=int(optimization_level))
    isa_ansatz = pm.run(ansatz)

    isa_observables = (
        qubit_hamiltonian.apply_layout(isa_ansatz.layout)
        if hasattr(qubit_hamiltonian, "apply_layout")
        else qubit_hamiltonian
    )

    opt = COBYLA(maxiter=maxiter, rhobeg=1.0, tol=1e-6)

    if x0 is None:
        rng = np.random.default_rng(seed)
        x0 = rng.random(isa_ansatz.num_parameters) if isa_ansatz.num_parameters > 0 else np.array([])

    energies: list[float] = []

    def cost(p: np.ndarray) -> float:
        ev = sampler_expectation_sparsepauliop(
            sampler=sampler,
            base_circuit=isa_ansatz,
            params=np.asarray(p, dtype=float),
            operator=isa_observables,
            shots=int(shots),
        )
        e = interpret_total_energy(ev, problem)
        energies.append(e)
        if verbose:
            print(f"Eval {len(energies):03d}: E = {e:.10f} Ha")
        return float(e)

    if isa_ansatz.num_parameters == 0:
        e = cost(x0)
        return {"fun": float(e), "x": np.array(x0, copy=True), "energies": energies}

    res = opt.minimize(fun=cost, x0=np.asarray(x0, dtype=float))
    return {"fun": float(res.fun), "x": np.array(res.x, copy=True), "energies": energies}
```

## Step 5: Build the Bond Scan function

This function performs the VQE function at each point of the bond length array and collects the energies to plot the potential energy surface of the H2 molecule.

```python
def bond_scan_h2_vqe_sampler(
    *,
    distances=None,
    reps: int = 5,
    maxiter: int = 120,
    seed: int = 42,
    warm_start: bool = True,
    basis: str = "sto3g",
    active_space: tuple[int, int] | None = (2, 2),
    sampler=None,
    shots: int = 4096,
    backend=None,
    optimization_level: int = 1,
):
    """  
    Performs a bond distance scan for H2 using VQE with shot-based sampling.
    At each distance, the VQE is run to find the ground state energy using
    the UCCSD ansatz and Jordan-Wigner mapping. The results are plotted to show
    the energy as a function of bond distance and the convergence behavior.
    """
    if distances is None:
        distances = np.linspace(0.54, 0.94, 5)

    distances = np.asarray(distances, dtype=float)
    energies = []
    trajectories = []
    params = []
    prev_x = None

    for i, d in enumerate(distances):
        Hq, problem = build_h2_qubit_hamiltonian_jw(
            float(d),
            basis=basis,
            charge=0,
            spin=0,
            freeze_core=False,
            active_space=active_space,
        )
        ansatz = build_h2_uccsd_ansatz(problem, reps=reps)

        x0 = prev_x if (warm_start and prev_x is not None and len(prev_x) == ansatz.num_parameters) else None

        out = run_vqe_single_point_sampler(
            problem=problem,
            qubit_hamiltonian=Hq,
            ansatz=ansatz,
            sampler=sampler,
            shots=shots,
            backend=backend,
            optimization_level=optimization_level,
            x0=x0,
            maxiter=maxiter,
            seed=seed,
            verbose=False,
        )

        e = float(out["fun"])
        energies.append(e)
        trajectories.append(list(out["energies"]))
        params.append(np.array(out["x"], copy=True))
        prev_x = params[-1]

        print(f"[{i+1:02d}/{len(distances):02d}] d={d:.4f} Å  E={e:.10f} Ha")

    energies = np.asarray(energies, dtype=float)

    # Figure 1 Bond Scan 
    plt.figure(figsize=(6, 4))
    plt.plot(distances, energies, marker="o")
    plt.xlabel("Bond distance (Å)")
    plt.ylabel("VQE total energy (Ha)")
    plt.title(f"H2 bond scan (Sampler, shots={shots}, UCCSD reps={reps})")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("h2_bond_scan_sampler.png", dpi=300)
    plt.show()
    # Figure 2 Convergence Plot
    # This plot displays the convergence of the VQE energy for each bond distance
    plt.figure(figsize=(6, 4))
    for i, traj in enumerate(trajectories):
        plt.plot(range(1, len(traj) + 1), traj, marker="o", label=f"d={distances[i]:.2f} Å")
    plt.xlabel("VQE Iteration")
    plt.ylabel("VQE total energy (Ha)")
    plt.title("VQE Convergence for H2 at Different Bond Distances")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("h2_vqe_convergence_sampler.png", dpi=300)
    plt.show()

    return {"distances": distances, "energies": energies, "convergence": trajectories, "optimal_params": params}
```

# Final Step

In this final step we run the bond scan function from above and save the plots. Again we calculated 5 points between 0.54Å and 0.94Å with 5 reps and a maximum of 20 iterations for the optimizer.

```python
res_scan = bond_scan_h2_vqe_sampler(
    distances=np.linspace(0.54, 0.94, 5),
    reps=5,
    maxiter=20,
    sampler=sampler,
    shots=SHOTS,
    backend=BACKEND,
    optimization_level=OPT_LEVEL,
)
res_scan["distances"], res_scan["energies"]
```

# Shot Count Helper

This last cell here is a helper cell to calculate the total number of shot count for the VQE algorithm. We first evaluate the number of measured Pauli-Terms, multiply this with the shots. Next we run the VQE algorithm for a max number of 20 iterations and get the total shot count for one specific bond length.

```python
def num_measured_pauli_terms(qubit_hamiltonian, *, tol: float = 1e-12) -> int:
    """
    Helper function to count the number of non-trivial Pauli terms
    in the qubit Hamiltonian that need to be measured.
    """

    labels = list(qubit_hamiltonian.paulis.to_labels())
    coeffs = np.asarray(qubit_hamiltonian.coeffs)
    n = 0
    for lab, c in zip(labels, coeffs):
        if np.isclose(np.real(c), 0.0, atol=tol):
            continue
        if set(lab) == {"I"}:
            continue
        n += 1
    return n

Hq, problem = build_h2_qubit_hamiltonian_jw(0.74)
n_terms = num_measured_pauli_terms(Hq)
shots_per_eval = SHOTS * n_terms
print("Measured Pauli terms:", n_terms)
print("Shots per energy evaluation:", shots_per_eval)

out = run_vqe_single_point_sampler(
    problem=problem,
    qubit_hamiltonian=Hq,
    ansatz=build_h2_uccsd_ansatz(problem, reps=5),
    sampler=sampler,
    shots=SHOTS,
    backend=BACKEND,
    optimization_level=OPT_LEVEL,
    maxiter=20,
)

n_evals = len(out["energies"])
total_shots_this_distance = shots_per_eval * n_evals
print("Cost evals:", n_evals)
print("Total shots (this distance):", total_shots_this_distance)
```
