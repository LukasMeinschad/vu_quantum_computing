# Helper Module to Compute Full-CI Energies for the respective Molecules

This can then be used as a proof of correctness for the VQE implementation.

```python
import pyscf
from pyscf import gto, scf, fci, mcscf
```

```python
def build_molecule_from_xyz(xyz_file, basis="sto-6g", spin=0, charge=0, symmetry=True):
    """  
    Helper function to build a PySCF molecule object from a xyz file

    Args:
        xyz_file (str): Path to the xyz file.
        basis (str): Basis set to use.
        spin (int): Spin multiplicity.
        charge (int): Charge of the molecule.
        symmetry (bool or str): Whether to use symmetry. If given a string of point
                                group name, the given point group symmetry will be used.
    """
    with open(xyz_file, 'r') as f:
        lines = f.readlines()

    # First line is number of atoms
    n_atoms = int(lines[0].strip())

    # Second line is comment line
    # Then we have atom lines of type: Element x y z
    atom_list = []
    for i in range(2, 2 + n_atoms):
        parts = lines[i].split()
        element = parts[0]
        x,y,z = float(parts[1]), float(parts[2]), float(parts[3])
        atom_list.append([element, (x,y,z)])

    # Build the molecule object
    mol = gto.Mole()
    mol.build(
        verbose=0, # Print symmetry details
        atom=atom_list,
        basis=basis,
        spin=spin,
        charge=charge,
        symmetry=symmetry
    )

    return mol
```

```python
h2_mol = build_molecule_from_xyz("/Users/lukas/Desktop/vu_quantum_computing/test_molecules/h2.xyz", basis="sto-3G", spin=0, charge=0, symmetry=True)
h2_hf = h2_mol.RHF().run()
h2_fci = fci.FCI(h2_hf).run()
print(f"Hartree-Fock Energy for H2: {h2_hf.e_tot:.6f} Ha")
print(f"Full-CI Energy for H2: {h2_fci.e_tot:.6f} Ha")
```
